```python
import numpy as np
```

```python
class LinearRegression:
    def __init__(self,learning_rate = 0.01, n_iter = 1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def fit(self,x, y): # x_train, y_train
        m, n = x.shape # No of feature, No of sample

        # step 1 --- Intialize
        self.bias = 0
        self.weights = np.zeros(n) #(0,0,0,0)


        # Griadient Descent
        for i in range(self.n_iter):
            
            #step 2 -- Calclate y_prediction
            y_pred = self.bias + np.dot(x, self.weights)
    
            #step 3 -- gradient
            db = (1/m) * sum(y_pred - y)
            dw = (1/m) * np.dot(x.T, (y_pred - y))
            
            # step 4 (apply convergence theorem -- params update
            self.bias = self.bias - self.lr * db
            self.weights = self.weights - self.lr * dw

    def predict(self,x):
        y_pred =  y_pred = self.bias + np.dot(x, self.weights)
        return y_pred
```

```python
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2,4,6,8,10])
model = LinearRegression()
model.fit(x, y)
y_pred = model.predict(x)
print(y_pred)


print(model.bias)
print(model.weights)
```
